**Description:** A fun Python-based Dare Game that generates personalized dares based on your current mood, a number (0-9), and an alphabet (A-F). Perfect for parties, friends, or breaking the ice!

## 🎮 How to Play

1. Run the code cell below.
2. Enter your mood (H/F/W/C/S/E).
3. Enter a number from 0-9.
4. Enter an alphabet from A to F.
5. Get a hilarious or meaningful dare!

The game covers moods like:
- **Happy (H)**
- **Funny (F)**
- **Weird (W)**
- **Crazy (C)**
- **Sad (S)**
- **Emotional (E)**



```python
## 📋 The Complete Python Code
print("<----WELCOME---->")
print("Dare Game")
print("How are you feeling Now ")
moods= input("Enter H for Happy, F for Funny, W for Wiered, C for Crazy, S for Sad and E for Emotional \n")
num= int(input("Enter Number from 0-9 \n"))
alphabet = input("Enter Aplabet from A to F \n")
alpha= alphabet.lower()
mood= moods.lower()

# Happy mood
if (mood == "h" and num in [0,5] and alpha== "a" or alpha == "c"):
    print("Give high-fives to 5 strangers.")
elif (mood == "h" and num in [2,7] and alpha== "b" or alpha == "e"):
    print("Compliment 3 people sincerely.")
elif (mood == "h" and num in [1,6] and alpha== "d" or alpha == "f"):
    print("Sing your favorite song loudly.")

# Funny mood
elif (mood == "f" and num in [1,6] and alpha== "b" or alpha == "d"):
    print("Dance publicly for 30 seconds.")
elif (mood == "f" and num in [3,8] and alpha== "a" or alpha == "f"):
    print("Walk like a penguin for 2 minutes.")
elif (mood == "f" and num in [4,9] and alpha== "c" or alpha == "e"):
    print("Laugh continuously for 30 seconds.")

# Weird Mood
elif (mood == "w" and num in [0,4] and alpha== "b" or alpha == "f"):
    print("Pretend you're an alien introducing yourself.")
elif (mood == "w" and num in [2,5] and alpha== "a" or alpha == "d"):
    print("Speak only in rhymes for 2 minutes.")
elif (mood == "w" and num in [7,9] and alpha== "c" or alpha == "e"):
    print("Walk backwards until your next turn.")

# Crazy Mood
elif (mood == "c" and num in [1,3] and alpha== "b" or alpha == "c"):
    print("Create a new dance move and perform it.")
elif (mood == "c" and num in [6,8] and alpha== "d" or alpha == "f"):
    print("Act like a chicken until someone laughs.")
elif (mood == "c" and num in [0,9] and alpha== "a" or alpha == "e"):
    print("Narrate everything you do dramatically.")

# Sad mood
elif (mood == "s" and num in [2,7] and alpha== "b" or alpha == "e"):
    print("Share a lesson you learned from a difficult day.")
elif (mood == "s" and num in [1,5] and alpha== "a" or alpha == "d"):
    print("Talk about someone you're grateful for.")
elif (mood == "s" and num in [3,8]  and alpha== "c" or alpha == "f"):
    print("Tell a meaningful childhood memory.")

# Emotional mood
elif (mood == "e" and num in [0,6] and alpha== "a" or alpha == "c"):
    print("Call or text someone and thank them.")
elif (mood == "h" and num in [2,9] and alpha== "b" or alpha == "d"):  # Note: Possible typo in original (h instead of e)
    print("Give everyone in the group one genuine compliment.")
elif (mood == "h" and num in [4,7] and alpha== "f" or alpha == "e"):  # Note: Possible typo in original
    print("Share your biggest dream for the future.")
else:
    print("Invalid Input")
```
