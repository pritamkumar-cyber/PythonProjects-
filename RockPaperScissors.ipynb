# Rock Paper Scissors Game

This is a simple Rock Paper Scissors game made in Python. The player chooses either Rock, Paper, or Scissors, and the computer randomly chooses one too.

The program then checks both choices and decides who wins. If both choose the same option, it is a tie. The game uses the normal rules where Rock beats Scissors, Scissors beats Paper, and Paper beats Rock.

I used `random.choice()` to make the computer's choice random and `if-elif-else` conditions to check the result.


```python
import random
```

```python
choices = ["Rock", "Paper", "scissors"]
player= input("Choose Rock, Paper, or Scissor \n").lower()
computer= random.choice(choices).lower()
print(f"Computer chose {computer}.")
if player==computer:
    print(f"It's a tie! Both chose {player}.")
elif(player=="rock" and computer== "scissors") or (player == "scissors" and computer == "paper") or (player== "paper" and computer == "rock"):
    print(f"You Win! {player} beats {computer}.")

else:
    print(f"Computer Wins! {computer} beats {player}.")
```
