# 💰 Expense Tracker Project

This is a simple **Expense Tracker** made using Python. It helps the user keep track of their daily expenses in an easy way.

The program allows the user to add an expense by entering the **date, category, description, and amount**. It also allows the user to view all the expenses they have added and check their **total spending**.

## 📌 Features

The Expense Tracker has four main options:

1. **Add Expense**  
   The user can add a new expense by entering the date, category, description, and amount.

2. **View All Expenses**  
   The user can see all the expenses that have been added to the tracker.

3. **View Total Spending**  
   The program calculates and displays the total amount spent.

4. **Exit**  
   The user can exit the Expense Tracker when they are done.

- `if-elif-else` statements
- User input using `input()`

- Adding items to a list using `append()`

- Calculating the total of expenses

## 📂 How the Data is Stored

Each expense is stored as a **dictionary** containing:

- **Date**
- **Category**
- **Description**
- **Amount**

All these expense dictionaries are then stored inside a list called `expensesList`.

## 🎯 Purpose of the Project

The main purpose of this project is to practice the basic concepts of Python by creating a small and useful real-life application.

It can be used to keep track of daily spending and understand where the money is being spent.


```python
#### expense Tracker Project

expenseList= []      #List of all expenses in the form of dictionary
print("Welcome to the Expense Tracker")

while True:
    print("===Menu===")
    print("1. Add expense")
    print("2. Veiw all expenses")
    print("3. View Total Khrcha")
    print("4. Exit")

    choice= int(input("Please enter your choice:"))

#Add Expense
    if(choice == 1):
        date= input("Kiss date pe kharcha kiya tha")
        category= input("kiss type ka kharcha kiya tha? (Food, Travel, Makeup, Party, Book and etc)")
        description= input("Write some description.")
        amount= float(input("Enter the amount: "))

        expense= {
            "date": date,
            "category": category,
            "description": description,
            "amount": amount
        }

        expenseList.append(expense)
        print(" \n Done Bro. Expense is added succesfully.")


#View All expenses
    elif(choice==2):
        if( len(expenseList)==0):
            print("No expense added.")
        else:
            print("===Your Expenses===")
            count=1
            for eachExpense in expenseList:
                print(f"Kharcha Number {count} -> {eachExpense["date"]}, {eachExpense["category"]}, {eachExpense["description"]}, {eachExpense["amount"]} ")
                count= count+1



#Viw total Spending
    elif(choice==3):
        total=0
        for eachExpense in expenseList:
            total= total+ eachExpense["amount"]

        print(" \n Total Kharcha: ", total)


#exit
    elif(choice==4):
        print("Thankyou for using our system")
        break

    else:
        print("Invalid Entry. Try Again")
        
```
