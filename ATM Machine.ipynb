### Atm machine with uses of oops concept

```python
class Atm:

    def __init__(self):
        self.pin = ""
        self.balance = 0
        self.menu()

    def menu(self):
        while True:
            user_input = int(input("""
            
        Enter 1 to Set PIN
        Enter 2 to Check Balance
        Enter 3 to Withdraw Money
        Enter 4 to Deposit Money
        Enter 5 to Cancel/Exit Transaction
        
        Enter your choice: """))

            if user_input == 1:
                self.create_pin()

            elif user_input == 2:
                self.check_balance()

            elif user_input == 3:
                self.withdraw_money()

            elif user_input == 4:
                self.deposit_money()

            elif user_input == 5:
                print("Transaction Cancelled")
                break

            else:
                print("Invalid Input")

    def create_pin(self):
        self.pin = input("Enter New PIN: ")
        print("PIN set successfully")

    def check_balance(self):
        pin = input("Enter PIN: ")

        if pin == self.pin:
            print("Your Balance is:", self.balance)
        else:
            print("Incorrect PIN")

    def withdraw_money(self):
        pin = input("Enter PIN: ")

        if pin == self.pin:
            amount = int(input("Enter withdrawal amount: "))

            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Withdrawal Successful")
                print("Remaining Balance:", self.balance)
            else:
                print("Insufficient Balance")

        else:
            print("Incorrect PIN")

    def deposit_money(self):
        pin = input("Enter PIN: ")

        if pin == self.pin:
            amount = int(input("Enter Amount: "))
            self.balance = self.balance + amount
            print("Deposited Successfully")
            print("Current Balance:", self.balance)

        else:
            print("Incorrect PIN")
```
