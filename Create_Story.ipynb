```python
def create_story():
    print("\nLet's create a story! Fill in the blanks")

    noun = input("Noun: ")
    verb= input("Verb: ")
    adjective= input("Adjective: ")
    place= input("Place: ")

    story= f"""
    Onhce upon a time, {adjective} {noun}
    decided to {verb} at the {place}.
    every one was surprised, but it turned out
    to be the best idea ever!
    """

    print("\nHere's your story:\n")
    print(story)

if __name__ == "__main__":
    create_story()
```
